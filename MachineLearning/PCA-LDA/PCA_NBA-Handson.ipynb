### **Importing the required packages**

```python
#Import Libraries
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

#Import ML Libraries
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, roc_auc_score
from sklearn.decomposition import PCA
```

**Reading and Exploring the data**

```python
#EDA
data = pd.read_csv('nba_final.csv')
```

```python
data = data[~(data.isnull().sum(axis=1)>5)]
```

```python
data.shape
```

```python
data.head()
```

```python
data['Pos2'] = data['Pos2'].fillna('None')
```

```python
data = data.fillna(0)
```

```python
data.isnull().sum()
```

```python
data.duplicated().sum()
```

```python
data.drop(columns=['Player.x', 'Player_ID'], inplace= True)
```

#### **Encoding the categorical columns**

```python
#Encoding
obj_cols = data.select_dtypes(object).columns
```

```python
obj_cols
```

```python
#Label Encoding
le = LabelEncoder()
for col in obj_cols:
    data[col] = le.fit_transform(data[col])
```

```python
data.head()
```

```python
data['Play'].value_counts()
```

### **Machine Learning Process**

```python
#Machine learn process
x = data.drop(columns='Play')
y = data['Play']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=10, stratify= y)
```

#### **Scaling or Mean Centering of the data**

```python
#Scaling
scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

#### **Apply the Logistic Regression on the data**

```python
#Apply Logistic Regression on the data
log_reg = LogisticRegression()
log_reg.fit(x_train_scaled, y_train)
```

```python
y_pred = log_reg.predict(x_test_scaled)
```

```python
accuracy_score(y_test, y_pred)
```

```python
roc_auc_score(y_test, y_pred)
```

```python
data.shape
```

### **Transform the data using PCA**

```python

pca = PCA(n_components=0.90)
x_train_pca = pca.fit_transform(x_train_scaled)
x_test_pca = pca.transform(x_test_scaled)

x_train_pca.shape
```

```python
pca.explained_variance_ratio_
```

#### **Apply Logistic Regression on the pca transformed data**

```python
log_reg2 = LogisticRegression()
log_reg2.fit(x_train_pca, y_train)
```

```python
y_pred2 = log_reg2.predict(x_test_pca)
```

```python
roc_auc_score(y_test, y_pred2)
```
